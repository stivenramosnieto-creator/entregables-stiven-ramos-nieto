*INTERPRETACION DE RESULTADOS
Squirrel es el tipo de animal con mayor numero de observaciones, con 403 registros,seguido de Hedgehog con 278. Esto indica que esta dos categorias concentran la mayor parte de las obcervaciones de el conjunto de datos.

2-European bison precenta el mayor peso promedio, con 588.40 kg muy por encima de lynx, cuyo promedio es de 23.27 kg. Esto muestra una diferencia considerable de tamaño y peso entre los tipos de animales analizados.

3-Poland representa el mayor peso promedio del pais, con 130.00 kg seguido de czech republic con74.50 kg y slovakia con 32.45 kg. eso indica que las observaciones registradas en estos paises incluyen, enpromedio, animales de mayor peso.

4-European Bison tambien presenta la mayor longitud corporal promedio, con 244.24 cm, mientras que lynx alcanza 82.90 cm. Esto es consistente con el mayor peso promedio del Europan bison y evidencia una diferencia marcada entre tamaño entre las especies.
