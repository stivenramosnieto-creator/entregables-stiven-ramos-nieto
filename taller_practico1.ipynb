```python
estudiantes = [("Ana", "bogota"), ("luis", "cucuta"), ("maria", "medellin"), ("carlos", "cali"), ("sofia", "bogota"), ("juan", "cucuta"), ("laura", "medellin"), ("pedro", "cartagena")]
for estudiante in estudiantes:
  nombre = estudiante[0]
  ciudad = estudiante[1]
  print(f"{nombre} viven en {ciudad}")
ciudades = {}
for estudiante in estudiantes:
    ciudad = estudiante[1]
    if ciudad in ciudades:
      ciudades[ciudad] = ciudades[ciudad] + 1
    else:
      ciudades[ciudad] = 1
print()
print("cantidad de estudiantes por ciudad:")
print(ciudades)
```
